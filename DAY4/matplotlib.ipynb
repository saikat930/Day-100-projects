# DATA VISULIZATION


```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = np.linspace(0,5,11)
y = x ** 10
y
```

```python
plt.plot(x,y)
plt.title("A plot")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,x)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(y,y)
```

```python
for i in range(1,25):
    plt.subplot(4,6,i)
```

```python
fig = plt.figure(figsize=(6,8))
axis2 = fig.add_axes([0.1,0.1,1,1])
axis2.plot(x,y)
axis1 = fig.add_axes([1.2,0.1,0.3,0.3])
axis1.plot(x,y)
```

```python
fig = plt.figure(figsize=(6,8))
axis2 = fig.add_axes([0.1,0.1,1,1])
axis2.scatter(x,y)
axis1 = fig.add_axes([1.2,0.1,0.3,0.3])
axis1.scatter(x,y)
axis3 = fig.add_axes([1.2,0.5,0.3,0.3])
axis3.scatter(x,y)


plt.savefig("fig1")
```
