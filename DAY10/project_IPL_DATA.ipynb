# **DATA VISUALIZATION PROJECT**


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('IPL.csv')
df.head()
```

1.Which team won the most matches?


```python
match_wins = df['match_winner'].value_counts()
match_wins
```

```python
sns.barplot(y=match_wins.index,x=match_wins.values, palette='viridis')
plt.title('Most match win by team')
plt.xlabel('Number of matches won')
```

2. Toss decision trends


```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title('toss decision trends')
```

3. Toss winner vs match winner


```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
print(f"percentage  = {(count/df.shape[0]*100).round(2)}")
```

4. How do teams win? (Runs vs Wickets)

3. Which bowler had the best bowling figures?


```python
df['runs_by_wicket'] = df['best_bowling_figure'].apply(lambda x:x.split('--')[1])
df['runs_by_wicket'] = df['runs_by_wicket'].astype(int)
```

```python
top_bowling_fig = df[df['highest_wickets'] == df['highest_wickets'].max()]
top_bowling_fig[top_bowling_fig['runs_by_wicket'] == top_bowling_fig['runs_by_wicket'].min()][['best_bowling','best_bowling_figure']]
```
