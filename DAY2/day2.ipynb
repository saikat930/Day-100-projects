```python
import numpy as np
import pandas as pd
```

```python
# reading the csv file and converting it into dataframe
df = pd.read_csv("C:/Users/SAIKAT/Downloads/employees_100 (1).csv")

# getting the information about the data frame
df.info()
```

```python
# getting na values from each column
df.isna().sum()
```

```python
# getting the sample of the data 
df.head()
```

```python
# fixing the na values if city ,age and salary
df['department'] = df['department'].str.strip().str.lower()
df['city'] = df['city'].fillna(df['city'].mode()[0])
df['department'] = df['department'].fillna(df['department'].mode()[0])
df['age']= df['age'].fillna(df['age'].mean())
df['age'] = df['age'].astype(int)
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
df['salary'] = df['salary'].fillna(df['salary'].mean())
df['salary'] = df['salary'].astype(int)

df.isnull().sum()
```

```python
df.info()
```

```python
# getting total number of duplicate rows
df.duplicated().sum()
```

```python
# finding duplicate employee_id
df.duplicated(subset=['employee_id']).sum()
```

```python
# fixing duplicate employee_id
df = df.drop_duplicates(subset=['employee_id'], keep='first')
df.duplicated(subset=['employee_id']).sum()
```

```python
# changing the datatype of joining_date column from str to datatime64
df['joining_date'] = pd.to_datetime(df['joining_date'])
df.info()
```

```python
# renaming the column names
df.rename(columns={'employee_id' : 'emp_id', 'experience' :"exp", 'department': 'dept','performance_score' : 'perf_score','name':'emp_name'},inplace=True)
df.head()
```

```python
# fixing column name structure
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")
df.columns.values
```

```python
# modifing and creating new columns (feature engineering)
df['annual_salary'] = df['salary'] * 12
def perf_category(perf):
    if perf >=8:
        return "Excellent"
    elif perf >=6:
        return "Good"
    else:
        return "Needs Improvement"

def categorize(num):
    if num > 5:
        return "Experienced"
    elif num <= 5:
        return "Learner"
    
df['perf_category'] = df['perf_score'].apply(perf_category)
df['experience_category'] = df['exp'].apply(categorize)
department_codes = {'data science': 'DS', 'engineering': 'ENG', 'marketing': 'MKT' ,'Finance': 'FIN', 'sales' : "SAL"}
df['department_code'] = df['dept'].map(department_codes)

df.head()
```

```python
# sorting and ranking
by_emp_id_sorted = df.sort_values(by='emp_id')
by_salary_sorted = df.sort_values(by='salary',ascending=False)
by_department_salary_sorted = df.sort_values(
    by=['dept','salary'],
    ascending = [True, False])

df['salary_rank'] = df['salary'].rank(ascending=False)
df.head(6)
```

```python
# grouping and aggregation
dept_city_salary_groupby = df.groupby(['dept','city'])[['salary']].mean()
dept_salary_perfscore_empid_groupby = df.groupby('dept').agg({
    'salary' : ["mean", "max"],
    'perf_score' : 'mean',
    'emp_id' : 'count'
})
department_summary = df.groupby('dept').agg(
    average_salary = ('salary', 'mean'),
    highest_salary = ('salary', 'max'),
    employee_count = ('emp_id', 'count')
)
department_summary
```
